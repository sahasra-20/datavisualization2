```python
import pandas as pd
import numpy as np

df=pd.read_csv("/kaggle/input/datasets/sobhanmoosavi/us-accidents/US_Accidents_March23.csv")

# Filter severity 4
df4 = df[df["Severity"] == 4].copy()

# Select relevant columns
columns = [
    "Severity",
    "Bump",
    "Crossing",
    "Junction",
    "Traffic_Signal"
]

parallel_df = df[columns].copy()

# Convert boolean columns to 0 and 1
for col in columns:
    parallel_df[col] = parallel_df[col].astype(int)

# Sample records
parallel_df = parallel_df.sample(
    n=min(10000, len(parallel_df)),
    random_state=42
)

parallel_df.to_csv("road_parallel.csv", index=False)

print(parallel_df.head())
```

```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler

# Select required attributes
weather_df = df[
    ["Severity","Temperature(F)", "Visibility(mi)",
     "Precipitation(in)", "Wind_Speed(mph)", "Humidity(%)"]
].copy()

# Remove missing values
weather_df = weather_df.dropna()

# Normalize weather attributes
weather_cols = [
    "Severity",
    "Temperature(F)", "Visibility(mi)",
     "Precipitation(in)", "Wind_Speed(mph)", "Humidity(%)"
]

scaler = MinMaxScaler()

weather_df[weather_cols] = scaler.fit_transform(
    weather_df[weather_cols]
)

# Sample records
weather_df = weather_df.sample(
    n=min(15000, len(weather_df)),
    random_state=42
)

weather_df.to_csv("weather_parallel.csv", index=False)

print(weather_df.head())
```
